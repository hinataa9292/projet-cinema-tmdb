```python
import os
import requests
from dotenv import load_dotenv
```

```python
load_dotenv()
token = os.getenv("TMDB_TOKEN")
print("Jeton chargé :", token is not None)
```

```python
url = "https://api.themoviedb.org/3/movie/550"   # 550 = Fight Club
headers = {"Authorization": f"Bearer {token}"}
params = {"language": "fr-FR"}

reponse = requests.get(url, headers=headers, params=params)
print("Code HTTP :", reponse.status_code)
```

```python
film = reponse.json()
print("Titre :", film["title"])
print("Année :", film["release_date"])
print("Note :", film["vote_average"])
```

```python
params = {"language": "fr-FR", "append_to_response": "credits"}
reponse = requests.get(url, headers=headers, params=params)
film = reponse.json()
print(film.keys())
```

```python
for acteur in film["credits"]["cast"][:5]:
    print(acteur["name"], "→", acteur["character"])
```

```python
for membre in film["credits"]["crew"]:
    if membre["job"] == "Director":
        print("Réalisateur :", membre["name"])
```

```python
print("Acteurs :", len(film["credits"]["cast"]))
print("Équipe technique :", len(film["credits"]["crew"]))
```

Les 10 premiers acteurs

```python
acteurs = film["credits"]["cast"][:10]
for a in acteurs:
    print(a["order"], a["name"])
```

Les postes importants de l'équipe

```python
POSTES = ["Director", "Screenplay", "Writer", "Producer", "Original Music Composer"]

equipe = [m for m in film["credits"]["crew"] if m["job"] in POSTES]
for m in equipe:
    print(m["job"], "→", m["name"])
```

Total après filtrage

```python
print("Personnes gardées :", len(acteurs) + len(equipe))
```
